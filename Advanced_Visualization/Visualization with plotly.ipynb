```python
fig = px.scatter(df , x='co2_tons_per_capita',
                 y='air_quality_index',
                 color='continent',
                 size='population_millions',
                 hover_name='city',
                 hover_data=['is_coastal','avg_temp_c'])
fig.show()
```

```python
df.columns
```

```python
px.histogram(df,x='air_quality_index',nbins=30,
             color='continent',
             marginal='box',
               title='Distribution of Air Quality Index by continent')
```

```python
fig = px.box(df , y='renewable_energy_share',
             x='continent',
             points='all')
fig.show()
```

```python
fig = px.violin(df , y='renewable_energy_share',
             x='continent',
             points='all',
             color='is_coastal')
fig.show()
```

```python
continent_counts = df['continent'].value_counts().reset_index()
continent_counts
```

```python
fig = px.bar(continent_counts , y='count',x='continent')
fig.show()
```

```python
ct = pd.crosstab(df['continent'],df['is_coastal']).reset_index()
ct
```

```python
fig = px.bar(pd.crosstab(df['continent'],df['is_coastal']),barmode='group')
fig.show()
```

```python
#another method 
dff = df.groupby(['continent','is_coastal']).size().reset_index(name='count')
dff
```

```python
fig = px.bar(dff ,x='continent',y='count',barmode='group',
             color='is_coastal')
fig.show()
```

```python
fig = px.scatter_geo(df , lon='longitude',lat='latitude',
                     color='continent',
                     projection='natural earth',
                     hover_name='city',
                     hover_data=['avg_temp_c','air_quality_index',
                               'is_coastal'])
fig.show()
```

```python
df.columns
```

```python
country_stats = df.groupby('iso_alpha3').agg({
    'population_millions':'sum',
    'co2_tons_per_capita':'mean',
    'air_quality_index':'mean',
    'renewable_energy_share':'mean'
}).reset_index()
```

```python
fig = px.choropleth( country_stats, locations='iso_alpha3',
                    color='population_millions',
                    projection='natural earth',
                    hover_data=['population_millions',
                               'air_quality_index' ])
fig.show()
```
